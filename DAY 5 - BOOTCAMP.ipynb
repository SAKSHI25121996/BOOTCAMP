## 🟢 Problem 2: Group + Sum

```python
data = [
 {"cust":"A","amount":100},
 {"cust":"B","amount":200},
 {"cust":"A","amount":50}
]
```

👉 Output:

```python
{"A":150, "B":200}
```

---

````python
# 👉 Output:
# {"A":150, "B":200}

# L1. Create empty dictionary (result = {})
# 2. Loop through the LIST
# 3. Access "cust" and "amount" from each dictionary
# 4. If customer NOT in result → Add customer with amount
# 5. If customer EXISTS in result → Add amount to existing value

data = [
 {"cust":"A","amount":100},
 {"cust":"B","amount":200},
 {"cust":"A","amount":50}
]
result = {}

for i in data:
    if i['cust'] not in result:
        result[i['cust']] = i['amount']
    else:
        result[i['cust']] += i['amount']
print(result)
````

## 🟢 Problem 4: JSON Flatten

```python
data = [
 {"id":1,"info":{"age":20,"city":"A"}},
 {"id":2,"info":{"age":25,"city":"B"}}
]
```

👉 Output:

```python
[
 {"id":1,"age":20,"city":"A"},
 {"id":2,"age":25,"city":"B"}
]
```

---

````python
# Your Thinking Steps:

# 1. Create empty result list
# 2. Loop through each dictionary in data
# 3. Get "id" from outer dictionary
# 4. Get "info" dictionary (contains age and city)
# 5. Create NEW flattened dictionary: {id, age, city}
# 6. Append this new dict to result list
````

````python
data = [
 {"id":1,"info":{"age":20,"city":"A"}},
 {"id":2,"info":{"age":25,"city":"B"}}
]

result = []

for i in data:
    if i['id'] not in result:
       for k,v in i['info'].items():
            if k == 'age':
                age = v
            if k == 'city':
                city = v
                result.append({'id': i['id'], 'age' : age , 'city': city})           
    
print(result)
````

````python
%sql
-- Create orders table with correct column names
CREATE OR REPLACE TABLE orders AS 
SELECT * FROM VALUES
(1, 'A', '2024-01-15', 100.00),
(2, 'A', '2024-02-10', 150.00),
(3, 'A', '2024-06-05', 200.00),
(4, 'B', '2024-01-20', 200.00),
(5, 'B', '2024-02-15', 150.00),
(6, 'B', '2024-03-10', 100.00),
(7, 'C', '2024-01-10', 100.00),
(8, 'C', '2024-02-20', 180.00),
(9, 'C', '2024-08-15', 120.00),
(10, 'D', '2024-01-25', 50.00),
(11, 'D', '2024-02-18', 75.00),
(12, 'D', '2024-03-22', 100.00),
(13, 'E', '2024-01-05', 50.00),
(14, 'E', '2024-01-28', 50.00),
(15, 'E', '2024-02-12', 80.00),
(16, 'E', '2024-02-25', 70.00),
(17, 'E', '2024-03-08', 120.00),
(18, 'E', '2024-03-20', 80.00)
AS t(order_id, customer_id, order_date, amount);
````

````python
🧠 Clear Thinking Steps (Follow This Exactly)
Step 1 → Get unique month per customer
Step 2 → Sort months
Step 3 → Compare with previous month
Step 4 → Check if difference = 1
Step 5 → Count streak = 3
⚠️ Important
Same month multiple orders → count once
Missing month → break streak
🎯 Expected Output
customer_id
````

````python
%sql
-- Alternative 2: Simpler approach with LEAD (looking forward)
WITH monthly_orders AS (
    SELECT DISTINCT
        customer_id,
        DATE_TRUNC('month', order_date) AS order_month
    FROM orders
),
with_next_months AS (
    -- Get the next 2 months for each record
    SELECT 
        customer_id,
        order_month AS month1,
        LEAD(order_month, 1) OVER (PARTITION BY customer_id ORDER BY order_month) AS month2,
        LEAD(order_month, 2) OVER (PARTITION BY customer_id ORDER BY order_month) AS month3
    FROM monthly_orders
)
-- Check if 3 consecutive months exist
SELECT DISTINCT customer_id
FROM with_next_months
WHERE 
    MONTHS_BETWEEN(month2, month1) = 1
    AND MONTHS_BETWEEN(month3, month2) = 1
ORDER BY customer_id;
````
