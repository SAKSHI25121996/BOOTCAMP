# AI Masterclass Notes: How Transformers Work

## Overview
This document explains the three-step process that happens behind the scenes when you interact with Large Language Models (LLMs) like GPT.

---

## Step 1: Tokenization
**What it is:**
* Every LLM has a vocabulary of tokens (words or word pieces)
* Each token has a unique token ID
* Your input text is broken down into these tokens

**Example:** "Hello world" might become tokens ["Hello", " world"] with IDs [1234, 5678]

---

## Step 2: Static Embeddings
**What it is:**
* Each token gets converted into a static embedding
* A static embedding is a vector (list of numbers) that represents the token
* This is a fixed representation for each token

**Example:** Token "Hello" might always map to vector [0.5, -0.2, 0.8, ...]

---

## Step 3: Contextual Embeddings
**What it is:**
* Tokens "talk" to their neighboring tokens to understand context
* This is called **bidirectional context**
* The same word can have different meanings based on surrounding words

**Key Concept:**
* **BERT Models** (Bidirectional Encoder Representations from Transformers) specialize in creating contextual embeddings
* BERT models are specifically designed to generate these embeddings
* **GPT Models** and most modern LLMs are built on these transformer principles

**Final Output:** A large vector like [0.1, -0.99, ..., 1] that contains contextual information

---

## How This Works in Practice

### When Using SDK/Client Packages
When you write simple code like:
```python
client.chat.completions.create(...)
```

**What happens behind the scenes:**
1. The client/SDK automatically handles tokenization
2. Creates static embeddings
3. Generates contextual embeddings
4. Sends the final vector to the LLM
5. The LLM understands your request because it has the full context

### Why Learn Manual Embedding Creation?

**Important for RAG Systems:**
* When building **RAG (Retrieval-Augmented Generation)** systems, you need to create contextual embeddings manually
* You'll need to understand this process to:
  * Store document embeddings in vector databases
  * Compare user queries with stored embeddings
  * Retrieve relevant context for your LLM

**The Trade-off:**
* **With SDK/Client:** Everything is automated, you don't see the complexity
* **For RAG:** You need to explicitly handle embeddings yourself

---

## Key Takeaway
The three-step process (Tokenization → Static Embeddings → Contextual Embeddings) is the foundation of how transformers work. Understanding this is essential for building advanced applications like RAG systems, even though client libraries handle it automatically for basic use cases.
