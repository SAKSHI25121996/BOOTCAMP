## What is LLM?

**LLM (Large Language Model)** is a type of language model.

A **Language Model** is an application that predicts the next word based on existing data.

**Example:** "The cat will sit on ___" → the model predicts the most likely next word.

---

## 3 Eras of Language Models

| Era | Approach |
| --- | --- |
| 1. Statistical | Traditional probability-based models (n-grams, etc.) |
| 2. Neural | Neural network-based models (RNNs, LSTMs) |
| 3. LLM | Transformer-based models (modern era) |

---

## The LLM Era

- In the LLM era, we work with **Transformers**.
- We keep the Transformer as the **base model architecture**.
- "LLM" is essentially a fancy/popular name for the **Transformer architecture** scaled to large data.
